## Question 1
 Create a function that will ask the user to input radius of a circle then compute the area if the area is > 20 display huge circle else small circle.

```python
def oac(radius):
  """ This is for calculating the area of a circle """
  area_circle = 3.14 * int(radius) ** 2
  print("Area of circle:", area_circle)
  if area_circle >20:
    print("Huge circle")
  else:
    print("Small circle")
```

```python
oac(20)
```

## Question 2
Provide a simple code using function that ask the user to input username and password, use the input function. Then if one of the input is wrong print the wrong input. For example, "username is wrong" or "password is wrong" or both. Print Welcome "username".

```python
correct_username = "admin"
correct_password = "1234"

def login(username, password):
  """ This is for account validation """
  if username == correct_username and password == correct_password:
        print(f"Welcome {username}!")
  elif username != correct_username and password != correct_password:
        print("Both username and password are wrong.")
  elif username != correct_username:
        print("Username is wrong.")
  else:
        print("Password is wrong.")
```

```python
login("keith", "ostan")
login("admin", "12345")
login("keith", "1234")
login("admin", "1234")
```

## Question 3
Write a function name div that will divide in1 and in2. Then if in2 = 0 raised error.

```python
def div(in1, in2):
    if in2 == 0:
        raise ZeroDivisionError("Cannot divide by zero.")
        print("Result:", div(in1, in2))
    return in1 / in2
```

```python
div(4,1)
```

## Question 4

Write a function name roomtemp with parameters temp then pass the value to conditional statement. If temp < 15, display "Lamig", if 16>temp > 23, display "Sakto lang", if temp <0, display "yelo yarn", if temp.> 40, display "BBQ"

```python
def roomtemp(temp):
    if temp < 0:
        print("yelo yarn")
    elif temp < 15:
        print("Lamig")
    elif 16 <= temp <= 23:
        print("Sakto lang")
    elif temp > 40:
        print("BBQ")
    else:
        print("Normal room temp")
```
